```python
import numpy as np
from projectEuler import Eratosthene

# La fonction suivante fournit le résultat s'il est <= N
def test(N):
    P = Eratosthene(N)
    for p in np.nonzero(P)[0][1:]:
        for k in range(1, int(np.sqrt(.5 * (N - p))) + 1):
            x = p + 2 * k * k
            P[x] = True

    x = 3
    while x <= N:
        if not P[x]:
            print(x)
            return True
        x += 2
    return False

N = 37
while not test(N):
    N *= 2
```
