```python
import numpy as np
from projectEuler import Eratosthene

N = 10 ** 6
t = Eratosthene(N)

S = []
for p in np.nonzero(t)[0][4:]:
    if t[p]:
        l = len(str(p)) - 1
        q =  p % 10 ** l
        ok = True
        while q > 0:
            if t[q]:
                l -= 1
                q %= 10 ** l
            else:
                ok = False
                break
        if ok:
            q = p // 10
            while q > 0:
                if t[q]:
                    q //= 10
                else:
                    ok = False
                    break
            if ok:
                S.append(int(p))
if len(S) == 11:
    print(sum(S))
else:
    raise Exception('Il faut augmenter N')
```
