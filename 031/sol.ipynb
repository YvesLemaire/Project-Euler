### Equation diophantienne linéaire (programmation dynamique)
Donnés $(a_0,\ldots,a_{n-1})\in \mathbf N^{*n}$ et $b\in\mathbf N$, soit à calculer le nombre de solutions $(x_0,\ldots,x_{n-1})\in \mathbf N^n$ de 
$$a_0x_0+\ldots+a_{n-1}x_{n-1}=b$$
Pour $i=0,\dots,b$ et $j=0,\ldots,n-1$, notons $N_{i,j}$ le nombre de solutions de $a_0x_0+\ldots+a_jx_j=i$. On a :

- $N_{i,0}=1$ si $a_0$ divise $i$ et $0$ sinon ;
- $N_{i,j}=N_{i,j-1}+N_{i-a_j,j}$ avec la convention $N_{k,h}=0$ si $k<0$, distinguer les solutions pour lesquelles $x_j=0$ des autres.

D'où le programme suivant qui calcule tous les $N_{i,j}$ (en commentaire, les invariants des deux boucles).

```python
def N(A, b):
    n = len(A)
    V = [0 if i % A[0] else 1 for i in range(b + 1)]
    for j in range(1, n):
        # V[i] = N[i, j - 1] pour i = 0,..,b
        for i in range(b + 1):
            # V[k] = N[k, j] si k = 0,..,i - 1 ; N[k, j - 1] si k = i,..,b
            if i >= A[j]:
                V[i] += V[i - A[j]]
            # V[k] = N[k, j] si k = 0,..,i ; N[k, j - 1] si k = i + 1,..,b
        # V[i] = N[i, j] pour i = 0,..,b
    return V[b]

N([1, 2, 5, 10, 20, 50, 100, 200], 200)
    
```
