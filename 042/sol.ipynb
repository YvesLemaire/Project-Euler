```python
with open('words.txt', 'r') as f:
    words = f.read().replace('"', '').split(',') # liste des mots
values = sorted([sum(ord(c) - 64 for c in w) for w in words]) # liste ordonnee des valeurs des mots
N = 0
n = 1
t = 1
# t = t_n
for v in values:
    while t < v:
        n += 1
        t += n
    # n minimum tq v <= t = t_n 
    if t == v: N += 1
N
```
