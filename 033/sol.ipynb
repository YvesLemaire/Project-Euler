```python
from math import gcd

curious_fractions = []

for p in range(1,9):
    for q in range(p + 1, 10):
        for r in range(1,10):
            P, Q = p + 10 * r, q + 10 * r
            if p * Q == P * q: curious_fractions.append((P,Q))
            P, Q = r + 10 * p, q + 10 * r
            if p * Q == P * q: curious_fractions.append((P,Q))
            P, Q = p + 10 * r, r + 10 * q
            if p * Q == P * q: curious_fractions.append((P,Q))
            P, Q = r + 10 * p, r + 10 * r
            if p * Q == P * q: curious_fractions.append((P,Q))
numerator, denominator = 1, 1
for P,Q in curious_fractions:
    numerator *= P
    denominator *= Q
denominator // gcd(numerator, denominator)
```
