Appelons solution un couple $(k,j)$, $1\leqslant j < k$ tel que $P_k-P_j$ et $P_k+P_j$ sont pentagonaux.

On cherche une solution minimale dans le sens où $P_k-P_j$ est minimum.

Cherchons dans un premier temps une solution :

```python
import math

p = lambda n: n * (3 * n - 1) // 2
ispentagonal = lambda v: p(round((math.sqrt(24 * v + 1) + 1) / 6)) == v 

def f():
    global k0, j0 
    k = 2
    pk = 5
    while True:
        j = k - 1
        pj = pk - 3 * k + 2
        while j > 0:
            if ispentagonal(pk - pj) and ispentagonal(pk + pj):
                k0, j0 = k, j
                print(pk - pj)
                return
            pj -= 3 * j - 2
            j -= 1 
        k += 1
        pk += 3 * k - 2

f()
```

Dans un deuxième temps, on vérifie que la solution trouvée $(k,j)=(k_0,j_0)=(2167, 1020)$ est minimale.

Si une solution $(k,j)\neq(k_0,j_0)$ satisfait $P_k-P_j<P_{k_0}-P_{j_0}$, alors
- $k>k_0$ car on a vu qu'il n'y avait pas de solution avec $k<k_0$ et il n'y en a pas de meilleure pour $k=k_0$ car on a traité les $j$ dans l'ordre décroissant, donc dans l'ordre croissant de $P_k-P_j$.
- $k< (P_{k_0}-P_{j_0} +2)/3$ car $3k-2=P_k-P_{k-1}\leqslant P_k-P_j<P_{k_0}-P_{j_0}$

Pour chaque $k$ dans l'intervalle défini par les deux conditions ci-dessus, on teste les $j<k$ tels que $P_k-P_j<P_{k_0}-P_{j_0}$.



```python
def g():
    pk = p(k0)
    pd0 = pk - p(j0)
    for k in range(k0 + 1, (pd0 + 2) // 3 + 1):
        j = k - 1
        pj = pk
        pk += 3 * k - 2 # pk = p(k)
        while pk - pj < pd0:
            if ispentagonal(pk - pj) and ispentagonal(pk + pj):
                print(f'Meilleure solution : {k=}, {j=}') 
            pj -= 3 * j - 2
            j -= 1 # pj = p(j)
            
g()
```
