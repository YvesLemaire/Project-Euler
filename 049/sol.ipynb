```python
import numpy as np
from itertools import permutations, combinations, combinations_with_replacement
from projectEuler import Eratosthene

t = Eratosthene(10 ** 4)

for u in combinations_with_replacement('0123456789', 4):
    L = []
    for v in permutations(u):
        p = int(''.join(v))
        if v[0] != '0' and t[p]:
            L.append(p)
    #print(L)
    for p, q, r in set(combinations(L, 3)):
        if 0 != r - q == q - p:
            print(f'{p}{q}{r}')
    
```
