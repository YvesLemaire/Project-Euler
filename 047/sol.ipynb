On maintient l'invariant suivant entre 2 variables $x$ et $\ell$ :

Chacun des $\ell$ entiers $x,x-1,\ldots,x-\ell+1$ admet $p$ facteurs premiers mais pas $x-\ell$.

```python
from projectEuler import primeDecomposition

p = 4

x = 1
l = 0
while l != p:
    x += 1
    l = l + 1 if len(primeDecomposition(x)) == p else 0
x - p + 1
```
