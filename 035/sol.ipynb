```python
import numpy as np
from projectEuler import Eratosthene

t = Eratosthene(10**6)

N = 0
for p in np.nonzero(t)[0]:
    if t[p]:
        s0 = str(p)
        s = s0[1:] + s0[0]
        ok = True
        while s != s0:
            if t[int(s)]:
                s = s[1:] + s[0]
            else:
                ok = False
                break
        if ok:
            N+= 1
N
```
