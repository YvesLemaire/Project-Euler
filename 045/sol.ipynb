```python
import math

t = lambda n: n * (n + 1) // 2

p = lambda n: n * (3 * n - 1) // 2
ispentagonal = lambda v: p(round((math.sqrt(24 * v + 1) + 1) / 6)) == v 

h = lambda n: n * (2 * n - 1)
ishexagonal = lambda v: h(round((math.sqrt(8 * v + 1) + 1) / 4)) == v 

n = 285
while True:
    n += 1
    T = t(n)
    if ispentagonal(T) and ishexagonal(T):
        print(T)
        break
```
