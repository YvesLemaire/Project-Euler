Appelons longueur d'un entier le nombre de chiffres de son écriture décimale. Posons

La constante de Champernowne est constituée :
- d'abord de la séquence de longueur $9$, $123456789$ formée en concaténant les $9$ entiers de longueur $1$, le rang du premier chiffre de la séquence étant $x_0=1$ ;
- puis de la séquence de longueur $2\times90$, $10111213\ldots9899$ concaténée des $90$ entiers de longueur $2$, le rang du premier chiffre est $x_1=x_0+9=10$ ;
- et ainsi de suite.

De manière générale, si $x_{n-1}$ est le rang du premier chiffre de la séquence de longueur 
$n\times 9\times10^{n-1}$, concaténée des $9\times10^{n-1}$ entiers de longueur $n$,
on a $x_n=x_{n-1}+n\times 9\times10^{n-1}$.

Ainsi, pour trouver le chiffre $d(r)$ situé au rang $r$, il faut principalement déterminer l'entier $n$ défini par $x_{n-1}\leqslant r<x_n$,
ce qui donne une complexité O(log( $r$ )).

La fonction suivante calcule $d(10^{10000})=7$ en 1 seconde.


```python
def d(r):
    n = 0
    x = 1 # x_0
    while x <= r:
        n += 1
        y = x # x_(n-1)
        x = x + n * 9 * 10 ** (n - 1) # x_n
    s = r - y 
    return int(str(10 ** (n - 1) + s // n)[s % n])

d(1) * d(10) * d(100) * d(1000) * d(10000) * d(100000) * d(1000000)
```
