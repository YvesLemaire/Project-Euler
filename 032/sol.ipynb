Soit $x \times y=z$ un produit pandigital avec $x\leqslant y$.

Notons $k,h$ et $n$ les nombres de chiffres des écritures décimales de $x,y$ et $z$.

On a :
- $k,h$ et $n$ sont $\geqslant1$ ;
- $k+h+n=9$ ;
- $10^{k+h-2}=10^{k-1}\times10^{h-1}\leqslant x\times y<10^{k}\times10^{h}=10^{k+h}$ ;
- $10^{n-1}\leqslant z=x\times y<10^{n}$.

Des deux dernières propriétés, on déduit que $k+h-1\leqslant n \leqslant k+h$.  
Donc, comme $n=k+h$ est impossible (il faudrait que $2n=9$), on a $n=k+h-1$.  
En reportant dans l'égalité $k+h+n=9$, il vient :

$k=1$ ou $2$, $k+h=5$ et $n=4$.

```python
from itertools import permutations

l = set()
for k in [1,2]:
    for p in permutations('123456789', 5):
        x = int(''.join(p[:k]))
        y = int(''.join(p[k:]))
        z = x * y
        sz = str(z)
        if len(sz) == 4 and set(p) | set(sz) == set('123456789'):
            print(f'{x} x {y} = {z}')
            l.add(z)
sum(l)
        
```
