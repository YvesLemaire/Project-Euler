$n\leqslant7$ car un nombre dont les chiffres sont $1,..,8$ ou $1,..,8,9$ est divisible par $3$ ($3$ divise $45=1+\ldots+8$ et $9$).

```python
from itertools import permutations
from sympy import isprime

m = 0
for n in range(2, 8):
    for t in permutations(range(1, n + 1), n):
        p = int(''.join(map(str,t)))
        if isprime((p)):
            m = max(m, p)
m
```
