O $((N/\ln N)^2)$

```python
import numpy as np 
from projectEuler import Eratosthene

N = 10**6

t = Eratosthene(N)
primes = np.nonzero(t)[0].tolist()
n = len(primes)
L = 0 
for k in range(n):
    h = k
    l = 1
    s = primes[k]
    while h < n - 1 and s <= N:
        if t[s] and l > L:
            L, K, S = l, k, s
        l += 1
        h += 1 # h = k + l - 1
        s += primes[h] # s = p_k + .. + p_h
print(f'Nombre de termes : {L}\nPremier terme : {primes[K]}\nSomme : {S}')
        
```
