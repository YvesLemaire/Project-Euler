```python
primes = [13, 11, 7, 5, 3, 2]
INT = lambda s: int(s[1:] if s[0] == '0' else s)
N = 0

def aux(s):
    global N
    l = len(s)
    for c in set('0123456789') - set(s):
        if l == 9:
            N += INT(c + s)
        else:
            if INT(c + s[:2]) % primes[l - 3] == 0:
                aux(c + s)

for n in range(17, 1000, 17):
    s = str(n) if n > 100 else '0' + str(n)
    if len(set(s)) == 3:
        aux(s)

N    
```
