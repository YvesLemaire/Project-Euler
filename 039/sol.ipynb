```python
from projectEuler import pythagoreanTriples

N = 1000
A = [0] * (N + 1)
for t in pythagoreanTriples(N // 2):
    p = sum(t)
    if p <= N:
        A[p] += 1

m = max(A)
for p in range(N + 1):
    if A[p] == m: 
         print(f'{m} triplets de perimetre {p}') 
```

#### Autre méthode de génération des triplets pythagoriciens

(plutôt moins efficace que la précédente ...)

La suite double $(a_k,b_k,b_k+h)$, $h\geqslant1,k\geqslant1$ décrite page 5 de [ce document](https://math.ou.edu/~dmccullough/teaching/slides/maa2003.pdf)
produit chaque $(a,b,c)$ où $a,b,c\geqslant1$ et $a^2+b^2=c^2$, exactement une fois.

Remarquer que les suites $a_k, b_k$ et $c_k$ sont croissantes pour un $h$ donné.

Voir aussi [ceci](http://www2.math.ou.edu/~dmccullough/teaching/pythagoras1.pdf) 

```python
from projectEuler import primeDecomposition

def pythagorean_triples(H, stop):
    """
    Engendre les triplets (a, b, c), 1 <= a < b < c, h = c - b < H, non stop((a, b, c)) et a^2 + b^2 = c^2
    Il faut que a <= a' et b <= b' et stop((a, b, c)) => stop((a', b', c'))
    Par exemple, pour obtenir tous les triplets tq a < b <= B, prendre H = (2^(1/2) - 1)B et stop = lambda t: t[1]  > B
    """
    for h in range(1, H):

        # Calcul de d
        D = primeDecomposition(h)
        pq = 1
        even = False
        for p, v in D:
            w = v % 2
            if p == 2 and w == 1:
                even = True
            pq *= p ** (w + v // 2)
        d = pq if even else 2 * pq

        const = d ** 2 // 2 // h
        a, b = h, 0
        while True:
            a, b = a + d, (d * a) // h + b + const
            c = b + h
            if stop((a, b, c)):
                break 
            else:
                if a < b:
                    yield (a, b, c)
```

Il est facile de vérifier que, pour tout triplet pythagoricien $(a,b,c)$ avec $a<b$ on a $\displaystyle h=c-b\leqslant\frac{1}{3\sqrt 3+1}(a+b+c)<0,17(a+b+c)$ 

```python
N = 1000
A = [0] * (N + 1)

for t in pythagorean_triples(round(.17  * N), lambda p: sum(p) > N):
    A[sum(t)] += 1

m = max(A)
for p in range(N + 1):
    if A[p] == m: 
         print(f'{m} triplets de perimetre {p}') 
```
