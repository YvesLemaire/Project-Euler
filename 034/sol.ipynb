Si un entier $n$ comporte $k$ chiffres, la somme des factorielles de ses chiffres est $\leqslant k\times 9!$.

De plus $10^{k-1}\leqslant n<10^{k}$. 

Donc, pour que $n$ soit un nombre curieux, il faut que $10^{k-1}\leqslant k\times 9!$, soit $k\leqslant7$. Ainsi, il faut $n\leqslant 7\times 9!$

```python
f = [1, 1, 2, 6, 24, 120, 720, 5040, 40320, 362880]
sum([n for n in range(10, 7 * f[9] + 1) if sum(f[int(x)] for x in str(n)) == n])
```
